```python
import docx
from docx.enum.table import WD_TABLE_ALIGNMENT
from docx.enum.text import WD_ALIGN_PARAGRAPH
from docx.oxml import parse_xml
from docx.oxml.ns import nsdecls
from docx.shared import Inches, Pt, RGBColor


def create_quotation():
    doc = docx.Document()

    for section in doc.sections:
        section.top_margin = Inches(0.75)
        section.bottom_margin = Inches(0.75)
        section.left_margin = Inches(0.75)
        section.right_margin = Inches(0.75)

    def set_cell_background(cell, fill_hex):
        tcPr = cell._element.get_or_add_tcPr()
        shd = parse_xml(f'<w:shd {nsdecls("w")} w:fill="{fill_hex}"/>')
        tcPr.append(shd)

    # Header
    title_p = doc.add_paragraph()
    title_p.alignment = WD_ALIGN_PARAGRAPH.RIGHT
    title_p.add_run("Ref: QUO/2026/DG-0908\nDate: 09/08/2026\n").font.size = Pt(
        9.5
    )

    h1_p = doc.add_paragraph()
    h1_run = h1_p.add_run("S. F Tower")
    h1_run.font.size = Pt(18)
    h1_run.font.bold = True
    h1_run.font.color.rgb = RGBColor(0x1F, 0x4E, 0x78)

    sub_p = doc.add_paragraph()
    sub_run = sub_p.add_run(
        "Agrabad Commercial Area, Chittagong\nMaterial list for P330 KVA DG"
        " installation ."
    )
    sub_run.font.size = Pt(12)
    sub_run.font.bold = True

    # Table
    col_widths = [
        Inches(0.7),
        Inches(3.4),
        Inches(0.7),
        Inches(0.8),
        Inches(1.1),
        Inches(1.1),
    ]
    table = doc.add_table(rows=1, cols=6)
    table.alignment = WD_TABLE_ALIGNMENT.CENTER

    hdr_titles = [
        "Sl. No",
        "Item Name & Size",
        "Unit",
        "Quantity",
        "Unit Price (BDT)",
        "Total Price (BDT)",
    ]
    for i, title in enumerate(hdr_titles):
        cell = table.rows[0].cells[i]
        cell.width = col_widths[i]
        p = cell.paragraphs[0]
        p.text = title
        p.alignment = (
            WD_ALIGN_PARAGRAPH.CENTER
            if i in [0, 2, 3]
            else (
                WD_ALIGN_PARAGRAPH.RIGHT
                if i in [4, 5]
                else WD_ALIGN_PARAGRAPH.LEFT
            )
        )
        p.runs[0].font.bold = True
        p.runs[0].font.color.rgb = RGBColor(0xFF, 0xFF, 0xFF)
        set_cell_background(cell, "1F4E78")

    items_data = [
        (1, "Electrical :-", "Power Cable (P-2, N-1) 1x95 rm", "Meter", 700),
        (2, "Electrical :-", "Control cable 2x1.2 rm flexible", "Meter", 500),
        (3, "Electrical :-", "Control cable 4x 1.2 rm flexible", "Meter", 200),
        (4, "Electrical :-", "Scheld cable 2x1.2 rm", "Meter", 100),
        (5, "Electrical :-", "Cable lugs 95 rm", "Nos", 56),
        (6, "Electrical :-", "Cable Tie 3x3x250mm, PVC", "Packet", 18),
        (7, "Electrical :-", "Cable Tie 3x3x100mm,PVC", "Packet", 3),
        (8, "Electrical :-", "PVC Tape R,Y,Blu,Black & G", "Nos", 35),
        (9, "Electrical :-", "PIB Tape", "Nos", 6),
        (10, "Electrical :-", "Cable Shoe 95rm, R,Y,Bl,Black", "Nos", 56),
        (11, "Electrical :-", "Cable Shoe 70 rm Green", "Nos", 12),
        (
            12,
            "Electrical :-",
            "Power cable cutting, laying & connection.",
            "Job",
            1,
        ),
        (13, "Earthing :-", "Earthing cable BYA 1x25 rm", "Meter", 30),
        (14, "Earthing :-", "Cable Socket 70 rm", "Nos", 48),
        (
            15,
            "Earthing :-",
            "Copper bus bar with supoter/ Isolator set",
            "Nos",
            4,
        ),
        (16, "Earthing :-", "Earthing boring cost", "Nos", 4),
        (17, "Earthing :-", "PVC tape Green", "Nos", 3),
        (18, "Mechanical :-", "M.S Pipe 6’’, 4mm thickness", "Ft", 50),
        (19, "Mechanical :-", "M.S Elbow 6’’", "Nos", 6),
        (20, "Mechanical :-", "Flange 6’’", "Nos", 8),
        (21, "Mechanical :-", "Gasket Klinger 3mm thk", "SFT", 4),
        (22, "Mechanical :-", "Nut & Bolts 16/75", "Nos", 48),
        (23, "Mechanical :-", "Washer 16/75", "Nos", 72),
        (24, "Mechanical :-", "Angle 50x50x6mm", "Ft", 40),
        (25, "Mechanical :-", "Cable Tray 70’x10’’x3’’", "Nos", 1),
        (26, "Mechanical :-", "Rawl bolt 12x100", "Nos", 60),
        (27, "Mechanical :-", "Red Oxide", "Ltr", 2),
        (28, "Mechanical :-", "Thineer", "Ltr", 2),
        (29, "Mechanical :-", "Paint", "Ltr", 2),
        (30, "Mechanical :-", "Radiator duct making & fitting", "Sft", 480),
        (31, "Mechanical :-", "Exhaust system silencer pipe work.", "Job", 1),
        (32, "Mechanical :-", "Battery charging 1250 electrolyte", "Nos", 2),
        (33, "Mechanical :-", "Busbar Junction Box 500 Amps4 Pole", "Nos", 1),
    ]

    current_cat = None
    for sl, cat, desc, unit, qty in items_data:
        if cat != current_cat:
            current_cat = cat
            cat_row = table.add_row()
            cat_cell = cat_row.cells[0]
            for c in range(1, 6):
                cat_cell.merge(cat_row.cells[c])
            p_cat = cat_cell.paragraphs[0]
            run_cat = p_cat.add_run(f"{cat}")
            run_cat.font.bold = True
            run_cat.font.color.rgb = RGBColor(0x1F, 0x4E, 0x78)
            set_cell_background(cat_cell, "D9E1F2")

        row = table.add_row()
        row.cells[0].paragraphs[0].text = f"{sl:02d}"
        row.cells[1].paragraphs[0].text = desc
        row.cells[2].paragraphs[0].text = unit
        row.cells[3].paragraphs[0].text = str(qty)

    doc.save("SF_Tower_Quotation.docx")


create_quotation()
```
