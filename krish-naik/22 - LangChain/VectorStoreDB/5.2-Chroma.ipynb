#### Chroma
Chroma is a AI-native open-source vector database focused on developer productivity and happiness. Chroma is licensed under Apache 2.0.

https://python.langchain.com/v0.2/docs/integrations/vectorstores/

```python
## building a sample vectordb

from langchain_chroma import Chroma
```

```python
from langchain_community.document_loaders import TextLoader
from langchain_community.embeddings import OllamaEmbeddings
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
## loading the data
loader = TextLoader("speech.txt")
data = loader.load()
```

```python
### splitting the text
splitter = RecursiveCharacterTextSplitter(chunk_size=1000,chunk_overlap=30)
splitted_data = splitter.split_documents(data)
```

```python
##### embedding
embedding = OllamaEmbeddings(model="nomic-embed-text")
vectordb = Chroma.from_documents(documents=splitted_data,embedding=embedding)
```

```python
### query it
query = "What does the speaker believe is the main reason the United States should enter the war?"

docs= vectordb.similarity_search(query)
docs
```

```python
### saving to thje disk
vectordb = Chroma.from_documents(documents=splitted_data,embedding=embedding,persist_directory="./chroma.db")
```

```python
#### load the disk

db2 = Chroma(persist_directory="./chroma.db",embedding_function=embedding)
```

```python
docs = db2.similarity_search(query)

## same for the similariy score,  similariry_search_with_score etc etc
```
