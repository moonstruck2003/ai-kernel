#### Getting started With Langchain And Open AI

In this quickstart we'll see how to:

- Get setup with LangChain, LangSmith and LangServe
- Use the most basic and common components of LangChain: prompt templates, models, and output parsers.
- Build a simple application with LangChain
- Trace your application with LangSmith
- Serve your application with LangServe

```python
#### loading the env files
import os
from dotenv import load_dotenv 
load_dotenv()
```

```python
os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")
## langsmith tracking
os.environ["LANGCHAIN_API_KEY"]=os.getenv("LANGCHAIN_API_KEY")
os.environ["LANGCHAIN_TRACING_V2"]="true"
os.environ["LANGCHAIN_PROJECT"]=os.getenv("LANGCHAIN_PROJECT")
```

```python
from langchain_groq import ChatGroq
llm = ChatGroq(
    model="openai/gpt-oss-20b",
    groq_api_key=os.getenv("GROQ_API_KEY")
)
print(llm)
```

```python
## input and get response from llm 
response = llm.invoke("What is Generatative AI?")
print(response)
```

```python
print(response.content)
```

```python
### prompt template (The LLM Behaviour)

from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages(
    [
        ("system","You are an expert AI Engineer. Provide me answers based on the question. "),
        ("user","{input}")
    ]
)
prompt
```

```python
### Chaining, combining the prompt template 
chain=prompt|llm

response=chain.invoke({"input":"can you tell me about langsmith?"})
print(response)
```

```python
### stroutput parser -> getting the LLM message and displaying it, we can create our own output parser

from langchain_core.output_parsers import StrOutputParser

output_parser = StrOutputParser()
chain = prompt|llm 

response = chain.invoke({"input":"Can you tell me about Langsmith?"})
```
